```python
def get_completion_from_messages(messages, model="gemini-3.6-flash", temperture=0, max_tokens=500):
    model=genai.GenerativeModel(model)

    # Gemini expects a single string input, so we concatenate all messages
    combined_text="\n" .join([msg["content"] for msg in messages])
#Take the content of every message and join them together, putting each message on a separate line.

    response=model.generate_content(combined_text)
    return response.text
```
